```python
import csv
import numpy as np
import matplotlib.pyplot as plt

# Lists to store data
mu = []
filling = []
delta_filling = []
scop = []
energy = []

# Read CSV file
with open("j2_-0.2_o5g3.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        mu.append(float(row["mu"]))
        filling.append(float(row["filling"]))
        delta_filling.append(float(row["delta_filling"]))
        scop.append(float(row["scop"]))
        energy.append(float(row["energy"]))

# Convert to numpy arrays (optional)
mu_2 = np.array(mu)
filling_2 = np.array(filling)
delta_filling_2 = np.array(delta_filling)
scop_2 = np.array(scop)
energy_2 = np.array([i*1e-2 for i in energy])


# Plotting
plt.plot(mu_2, filling_2, ".", color="red")
plt.plot(mu_2, filling_2, label=r'$n_{\sigma}$', color="red")
plt.plot(mu_2, delta_filling_2, ".", color="black")
plt.plot(mu_2, delta_filling_2, label=r'$\delta_{n_{\sigma}}$', color="black")
plt.plot(mu_2, scop_2, ".", color="blue")
plt.plot(mu_2, scop_2, label=r'$\Delta_m$', color="blue")
plt.grid()
plt.show()

mask = (mu_2 != -3.4)
plt.plot(mu_2[mask], energy_2[mask], ".", color="green")
plt.plot(mu_2[mask], energy_2[mask], label=r'$E$', color="green")

plt.grid()
plt.xlabel('Chemical Potential (mu)')
plt.legend()
plt.axhline(y=0.5, color='r', linestyle='--')  # horizontal line
# plt.savefig('t2_0.4_o5g3.png')
plt.show()

```

```python
import csv
import numpy as np
import matplotlib.pyplot as plt

# Lists to store data
mu = []
filling = []
delta_filling = []
scop = []
energy = []

# Read CSV file
with open("j2_-0.4_o5g3.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        mu.append(float(row["mu"]))
        filling.append(float(row["filling"]))
        delta_filling.append(float(row["delta_filling"]))
        scop.append(float(row["scop"]))
        energy.append(float(row["energy(e+2)"]))

# Convert to numpy arrays (optional)
mu_4 = np.array(mu)
filling_4 = np.array(filling)
delta_filling_4 = np.array(delta_filling)
scop_4 = np.array(scop)
energy_4 = np.array(energy)


# Plotting
plt.plot(mu_4, filling_4, ".", color="red")
plt.plot(mu_4, filling_4, label=r'$n_{\sigma}$', color="red")
plt.plot(mu_4, delta_filling_4, ".", color="black")
plt.plot(mu_4, delta_filling_4, label=r'$\delta_{n_{\sigma}}$', color="black")
plt.plot(mu_4, scop_4, ".", color="blue")
plt.plot(mu_4, scop_4, label=r'$\Delta_m$', color="blue")
plt.show()

plt.plot(mu_4, energy_4, ".", color="green")
plt.plot(mu_4, energy_4, label=r'$E$', color="green")

plt.grid()
plt.xlabel('Chemical Potential (mu)')
plt.legend()
plt.axhline(y=0.5, color='r', linestyle='--')  # horizontal line
# plt.savefig('t2_0.4_o5g3.png')
plt.show()
```
